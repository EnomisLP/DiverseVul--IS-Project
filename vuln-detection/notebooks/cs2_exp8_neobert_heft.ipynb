## 1. Dependencies

```python
# =============================================================================
# ONE-CELL PYREFT ENVIRONMENT SETUP (FIXED VERSION)
# =============================================================================
#
# Target environment:
#   Python       3.10
#   PyTorch      2.5.1 + CUDA 12.1
#   GPU          NVIDIA A100 80GB
#   Transformers 4.45.1       <- required by pyReFT 0.1.0
#   PEFT         0.13.2
#   pyReFT       0.1.0
#   xFormers     0.0.28.post3 <- matches PyTorch 2.5.1
#
# nnsight:
#   NOT installed.
#
# TensorFlow / Keras:
#   NOT used.
#
# IMPORTANT:
#   Run this cell in a fresh notebook runtime.
#   If packages are changed, this cell asks for ONE kernel restart.
#   After restart, run THIS SAME CELL again.
# =============================================================================

import sys
import os
import subprocess
import importlib
import importlib.metadata as metadata
import time

PYTHON = sys.executable

TARGETS = {
    "torch": "2.5.1",
    "torchvision": "0.20.1",
    "torchaudio": "2.5.1",
    "transformers": "4.45.1",
    "peft": "0.13.2",
    "xformers": "0.0.28.post3",
    "pyvene": "0.1.8",  # Critical pyreft dependency (latest available version)
    "pyreft": "0.1.0",
}

print("=" * 80)
print("ONE-CELL PYREFT ENVIRONMENT SETUP")
print("=" * 80)
print("Python:", PYTHON)
print("Version:", sys.version.split()[0])

# -----------------------------------------------------------------------------
# Helper functions
# -----------------------------------------------------------------------------

def installed_version(package):
    """Get installed version of a package."""
    try:
        return metadata.version(package)
    except metadata.PackageNotFoundError:
        return None


def strip_build_metadata(version_str):
    """
    Strip build metadata (e.g., '+cu121', '.post3') from version strings.
    
    Examples:
        "2.5.1+cu121" -> "2.5.1"
        "0.20.1+cu121" -> "0.20.1"
        "0.0.28.post3" -> "0.0.28"
    """
    if version_str is None:
        return None
    # Remove everything after '+' (CUDA suffix)
    version_str = version_str.split('+')[0]
    # Remove .postN suffix
    if '.post' in version_str:
        version_str = version_str.split('.post')[0]
    return version_str


def versions_match(current, target):
    """
    Compare two version strings, ignoring build metadata.
    
    Args:
        current: Currently installed version (e.g., "2.5.1+cu121")
        target: Target version (e.g., "2.5.1")
    
    Returns:
        True if base versions match, False otherwise
    """
    if current is None:
        return False
    
    current_clean = strip_build_metadata(current)
    target_clean = strip_build_metadata(target)
    
    return current_clean == target_clean


def pip_install(args, description):
    """Execute pip install command."""
    print("\n" + "=" * 80)
    print(description)
    print("=" * 80)

    cmd = [PYTHON, "-m", "pip", "install", "--no-cache-dir"] + args

    print("$", " ".join(cmd))

    result = subprocess.run(
        cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
    )

    print(result.stdout)

    if result.returncode != 0:
        raise RuntimeError(
            f"\n❌ Installation failed:\n{description}\n"
            f"Exit code: {result.returncode}"
        )


# =============================================================================
# STEP 1 — Check whether target environment is already installed
# =============================================================================

print("\n" + "=" * 80)
print("CHECKING EXISTING ENVIRONMENT")
print("=" * 80)

needs_install = False

for package, target in TARGETS.items():
    current = installed_version(package)

    if versions_match(current, target):
        print(f"✅ {package:<15} {current}")
    else:
        print(f"⚠️  {package:<15} current={current!r} target={target}")
        needs_install = True


# =============================================================================
# STEP 2 — Install coherent PyTorch + CUDA stack (ONLY IF NEEDED)
# =============================================================================

if needs_install:

    print("\n" + "=" * 80)
    print("INSTALLING COHERENT PYTORCH / CUDA STACK")
    print("=" * 80)

    pip_install(
        [
            "--index-url",
            "https://download.pytorch.org/whl/cu121",
            "torch==2.5.1",
            "torchvision==0.20.1",
            "torchaudio==2.5.1",
        ],
        "PyTorch 2.5.1 + CUDA 12.1",
    )

    # -------------------------------------------------------------------------
    # Core scientific / HF packages
    #
    # Keep these deliberately conservative.
    # -------------------------------------------------------------------------

    pip_install(
        [
            "numpy<2.1",
            "pandas",
            "scikit-learn",
            "matplotlib",
            "pyarrow",
            "joblib",
            "tqdm",
            "psutil",
            "accelerate",
            "einops",
            "sentencepiece",
        ],
        "Core scientific + Hugging Face dependencies",
    )

    # -------------------------------------------------------------------------
    # pyReFT's declared Transformers version
    # -------------------------------------------------------------------------

    pip_install(
        [
            "transformers==4.45.1",
            "peft==0.13.2",
        ],
        "Transformers / PEFT versions compatible with pyReFT",
    )

    # -------------------------------------------------------------------------
    # pyvene (critical pyreft dependency - MUST come before pyreft)
    #
    # pyvene is a hard dependency of pyreft that provides intervention/editing
    # capabilities. It MUST be installed before pyreft is imported.
    # -------------------------------------------------------------------------

    pip_install(
        [
            "pyvene==0.1.8",  # Latest available version compatible with pyreft 0.1.0
        ],
        "pyvene 0.1.8 (pyReFT critical dependency)",
    )

    # -------------------------------------------------------------------------
    # pyReFT
    #
    # --no-deps is intentional:
    # We already control the critical dependencies above (including pyvene).
    # We do not want pyReFT to trigger another large dependency-resolution
    # cascade that might conflict with our carefully controlled environment.
    # -------------------------------------------------------------------------

    pip_install(
        [
            "--no-deps",
            "pyreft==0.1.0",
        ],
        "pyReFT 0.1.0 (with --no-deps, pyvene already installed above)",
    )

    # -------------------------------------------------------------------------
    # xFormers
    #
    # --no-deps is CRITICAL.
    #
    # We already have the exact PyTorch version that xFormers 0.0.28.post3
    # was built for. Do not let pip replace it.
    # -------------------------------------------------------------------------

    pip_install(
        [
            "--no-deps",
            "xformers==0.0.28.post3",
        ],
        "xFormers 0.0.28.post3",
    )

    # -------------------------------------------------------------------------
    # Fix datasets/fsspec compatibility if datasets already exists.
    #
    # This prevents the conflict seen in your previous diagnostic.
    # -------------------------------------------------------------------------

    if installed_version("datasets") is not None:
        pip_install(
            [
                "fsspec[http]<=2026.6.0,>=2023.1.0",
            ],
            "Align fsspec with installed datasets",
        )

    print("\n" + "=" * 80)
    print("INSTALLATION COMPLETE")
    print("=" * 80)

    print(
        """\
⚠️  ONE KERNEL RESTART IS REQUIRED.

The packages above were installed/changed by pip.
Restart the notebook kernel now.

Then run THIS SAME CELL again.

The second run will detect the correct versions and perform
the final verification instead of reinstalling everything.
"""
    )

    raise SystemExit


# =============================================================================
# STEP 3 — Environment is already aligned
# =============================================================================

print("\n" + "=" * 80)
print("ENVIRONMENT ALREADY ALIGNED")
print("=" * 80)

for package, target in TARGETS.items():
    current = installed_version(package)

    if not versions_match(current, target):
        raise RuntimeError(
            f"❌ {package} mismatch: expected {target}, found {current}"
        )

    print(f"✅ {package:<15} {current}")


# =============================================================================
# STEP 4 — Explicitly configure this as a PyTorch-only environment
# =============================================================================

# These are only relevant to this Python process.
# They prevent Transformers from attempting unnecessary TF/Flax initialization.

os.environ["USE_TF"] = "0"
os.environ["USE_FLAX"] = "0"


# =============================================================================
# STEP 5 — Import verification
# =============================================================================

print("\n" + "=" * 80)
print("IMPORT VERIFICATION")
print("=" * 80)

failures = []


def check_import(name):
    try:
        module = importlib.import_module(name)
        version = getattr(module, "__version__", "unknown")
        print(f"✅ {name:<18} {version}")
        return module
    except ModuleNotFoundError as e:
        print(f"❌ {name:<18} ModuleNotFoundError: {e}")
        print(f"   → A required module is missing. Re-run Cell 2 after kernel restart.")
        failures.append((name, e))
        return None
    except ImportError as e:
        print(f"❌ {name:<18} ImportError: {e}")
        print(f"   → A dependency failed to import. Check the full error below.")
        failures.append((name, e))
        return None
    except Exception as e:
        print(f"❌ {name:<18} {type(e).__name__}: {e}")
        failures.append((name, e))
        return None


torch = check_import("torch")
torchvision = check_import("torchvision")
torchaudio = check_import("torchaudio")
transformers = check_import("transformers")
peft = check_import("peft")
accelerate = check_import("accelerate")
pyreft = check_import("pyreft")
xformers = check_import("xformers")


# =============================================================================
# STEP 6 — CUDA verification
# =============================================================================

print("\n" + "=" * 80)
print("CUDA / GPU VERIFICATION")
print("=" * 80)

if torch is None:
    raise RuntimeError("❌ PyTorch could not be imported.")

print("PyTorch       :", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("CUDA runtime  :", torch.version.cuda)

if not torch.cuda.is_available():
    raise RuntimeError(
        "❌ CUDA is not available. Enable the NVIDIA GPU runtime."
    )

print("GPU           :", torch.cuda.get_device_name(0))
print("GPU count     :", torch.cuda.device_count())

# Basic CUDA tensor operation
try:
    a = torch.tensor(
        [1.0, 2.0, 3.0],
        device="cuda",
        dtype=torch.float32,
    )

    b = a * 2

    torch.cuda.synchronize()

    print("CUDA tensor   : PASSED")
    print("Result        :", b.cpu().tolist())

    del a, b

except Exception as e:
    raise RuntimeError(
        f"❌ CUDA tensor test failed:\n{e}"
    )


# =============================================================================
# STEP 7 — xFormers CUDA verification
# =============================================================================

print("\n" + "=" * 80)
print("XFORMERS CUDA VERIFICATION")
print("=" * 80)

if xformers is None:
    raise RuntimeError("❌ xFormers import failed.")

try:
    from xformers.ops import memory_efficient_attention

    q = torch.randn(
        1, 4, 8, 64,
        device="cuda",
        dtype=torch.float16,
    )

    k = torch.randn(
        1, 4, 8, 64,
        device="cuda",
        dtype=torch.float16,
    )

    v = torch.randn(
        1, 4, 8, 64,
        device="cuda",
        dtype=torch.float16,
    )

    output = memory_efficient_attention(q, k, v)

    torch.cuda.synchronize()

    print("xFormers ops : IMPORT OK")
    print("CUDA kernel  : PASSED")
    print("Output shape :", tuple(output.shape))

    del q, k, v, output
    torch.cuda.empty_cache()

except Exception as e:
    raise RuntimeError(
        f"❌ xFormers CUDA operation failed:\n{e}"
    )


# =============================================================================
# STEP 8 — pyReFT verification
# =============================================================================

print("\n" + "=" * 80)
print("PYREFT VERIFICATION")
print("=" * 80)

if pyreft is None:
    raise RuntimeError("❌ pyReFT import failed.")

print("pyReFT module :", pyreft.__file__)
print("pyReFT        : IMPORT OK")


# =============================================================================
# STEP 9 — typing_extensions verification (optional check)
# =============================================================================

print("\n" + "=" * 80)
print("TYPING EXTENSIONS")
print("=" * 80)

try:
    import typing_extensions
    
    print("File          :", typing_extensions.__file__)
    
    # Sentinel was added in typing_extensions 4.6.0+
    # It's nice to have but not critical for pyreft
    sentinel_available = hasattr(typing_extensions, "Sentinel")
    
    if sentinel_available:
        print("Sentinel      : Available ✅")
    else:
        print("Sentinel      : Not available (OK - not critical)")
    
    print("typing_extensions: OK")

except Exception as e:
    print(f"⚠️  typing_extensions warning: {e}")
    print("   (This is not critical and can be ignored)")


# =============================================================================
# STEP 10 — Dependency check
# =============================================================================

print("\n" + "=" * 80)
print("PIP DEPENDENCY CHECK")
print("=" * 80)

result = subprocess.run(
    [PYTHON, "-m", "pip", "check"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
)

if result.stdout.strip():
    print(result.stdout.strip())
else:
    print("✅ pip check reports no dependency conflicts")


# =============================================================================
# STEP 11 — Explicitly explain nnsight
# =============================================================================

print("\n" + "=" * 80)
print("NNSIGHT STATUS")
print("=" * 80)

nnsight_version = installed_version("nnsight")

if nnsight_version is None:
    print("ℹ️  nnsight is intentionally NOT installed.")
    print()
    print(
        "This is deliberate. Your previous nnsight 0.7.0 installation "
        "was incompatible with the Transformers 4.45.1 stack because "
        "nnsight attempted to import AutoModelForImageTextToText."
    )
else:
    print(f"⚠️  nnsight {nnsight_version} is installed.")
    print(
        "It is not required for the core pyReFT + Transformers + "
        "xFormers environment and is intentionally not part of this setup."
    )


# =============================================================================
# FINAL RESULT
# =============================================================================

if failures:
    print("\n" + "=" * 80)
    print("❌ ENVIRONMENT FAILED")
    print("=" * 80)

    for name, error in failures:
        print(f"\n{name}:")
        print(f"  {type(error).__name__}: {error}")

    raise RuntimeError(
        "One or more required imports failed. "
        "See the diagnostics above."
    )


print("\n" + "=" * 80)
print("🎉 ENVIRONMENT READY")
print("=" * 80)

print(f"Python        : {sys.version.split()[0]}")
print(f"PyTorch       : {torch.__version__}")
print(f"CUDA          : {torch.version.cuda}")
print(f"GPU           : {torch.cuda.get_device_name(0)}")
print(f"Transformers  : {transformers.__version__}")
print(f"PEFT          : {peft.__version__}")
print(f"pyReFT        : {installed_version('pyreft')}")
print(f"xFormers      : {xformers.__version__}")

print()
print("✅ PyTorch + CUDA")
print("✅ Transformers")
print("✅ PEFT")
print("✅ pyReFT")
print("✅ xFormers")
print("✅ CUDA tensor operation")
print("✅ xFormers CUDA attention")
print("✅ typing_extensions")

print()
print("You now have a clean PyTorch/pyReFT environment.")
print("Do NOT install nnsight or tf-keras unless your actual project requires them.")
print("=" * 80)
```
